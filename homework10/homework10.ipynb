```python
cols_to_skip = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df.columns:
    if col not in cols_to_skip:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.dtypes
```

```python
# the 3 different types of data are object, float64, int64
# 'Planet', 'Global Magnetic Field?', and 'Ring System?' are still objects because they are the only values that are not numerical values 
```

```python
def attach_units(old_col, new_col, unit):
    new_values = []
    for val in df[old_col]:
        new_values.append(val * unit)
    
    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)
```

```python
attach_units("Distance from Sun (10^6 km)", "Distance from Sun (km)", 1e6 * u.km)
```

```python
km_cols = [col for col in df.columns if 'km' in col]

for col in km_cols:
    try:
        cleaned = (
            df[col]
            .astype(str)               
            .str.replace(',', '')      
            .str.replace(' km', '', case=False)  
            .str.strip()           
        )

        values = pd.to_numeric(cleaned, errors='coerce').to_numpy(dtype=float)

        df[col] = (values * u.km).to(u.AU).value

        df.rename(columns={col: col.replace('km', 'AU')}, inplace=True)

    except Exception as e:
        print(f"Skipping {col} due to error: {e}")

df.head()
```

```python
M = df['Mass (10^24kg)'].values * 1e24 * u.kg

diam_au = np.array(df['Diameter (AU)']).astype(float)

if diam_au.ndim == 2:
    diam_au = diam_au[:, 0]

R = (diam_au / 2) * u.AU
R = R.to(u.m)

print(M.shape)
print(R.shape)
```

```python
g = (const.G * M / R**2).to(u.m / u.s**2)
df['Surface Gravity (m/s^2)'] = g.value
```
