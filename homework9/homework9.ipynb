```python
x = df['x']
y = df['y_noise_1']
plt.figure()
plt.scatter(x,y, label= "Noisy Data", color = 'gray')
plt.xlabel= ('x')
plt.ylabel= ('y_noise_1')
plt.legend()
plt.title('Noisy Data')
plt.show
```

```python
def model(x, A, k):
    return A * np.exp(-k * x)

p0 = [1, 1]
popt, pcov = curve_fit(model, x, y_noise_1, p0=p0)

A_fit, k_fit = popt

print(f"Fit: y = {A_fit:.2f} * e^(-{k_fit:.2f} * x)")

residuals = y_noise_1 - model(x, A_fit, k_fit)
chi2 = np.sum(residuals**2)
dof = len(x) - len(popt)
chi2_reduced = chi2 / dof

plt.scatter(x, y_noise_1, label="Noisy data")
plt.plot(x, model(x, A_fit, k_fit), "r-", label="Fitted model")
plt.plot(x, model(x, 5, 0.8), "k--", label="True model")  
plt.title(f"Noise=1, Reduced Chi^2={chi2_reduced:.2f}")
plt.legend()
plt.show()
```

```python
def fit_model(x, y):
    p0 = [1, 1]
    popt, pcov = curve_fit(model, x, y, p0=p0)
    A_fit, k_fit = popt
    
    residuals = y - model(x, A_fit, k_fit)
    chi2 = np.sum(residuals**2)
    dof = len(x) - len(popt)
    chi2_reduced = chi2 / dof
    
    return A_fit, k_fit, chi2_reduced



noisy_datasets = [y_noise_1, y_noise_2, y_noise_3, y_noise_4, y_noise_5]
noise_levels = [1, 2, 3, 4, 5]

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

for ax, noise, y_noise in zip(axs.flat, noise_levels[:4], noisy_datasets[:4]):
    A_fit, k_fit, chi2_reduced = fit_model(x, y_noise)
    
    ax.scatter(x, y_noise, color="gray", s=15, label="Noisy data")
    ax.plot(x, model(x, A_fit, k_fit), "r-", label="Fitted model")
    ax.plot(x, model(x, 5, 0.8), "k--", label="True model")
    
    ax.set_title(f"Noise={noise}, Reduced χ²={chi2_reduced:.2f}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.legend()

plt.tight_layout()
plt.show()
```
