```python
# different discovery methods: transit (388), radial velocity (104), microlensing (61), imaging (5), transit timing variations (3), eclipse timing variations (2), disk kinematics (1)
# Transit has the most (388), disk kinematics has the least (1)
# ---Descritpions---
# Transit: A star's brightness is monitored and changes are observed. When an exoplanet passes infront of the star, it temporarily dims

# Radial Velocity: Measures how much the star wobbles due to the gravitational pull of the exoplanet

# Microlensing: Observes a temporary brightening of a distant star when a closer star, and its planet(s) passes in front of it

# Imaging: directly finds exoplanets by taking pictures of them

# Transit Timing Variations: measuring  changes in the timing of a known planet's transit across its star due to gravitational pull of exoplanet

# Eclipse Timing Variations: detects exoplanets by measuring tiny changes in when two stars eclipse each other, caused by an exoplanets planet’s gravity making their orbits wobble

# Disk Kinematics: Observing the movement of gas and dust in a protoplanetary disk caused by the gravitational pull of an exoplanet
```

```python
# plt.figure(figsize=(8,6))

# discoverymethod_list = df["discoverymethod"].unique()

# for discoverymethod in discoverymethod_list:
#     df_subset=df[df["discovermethod"]==discoverymethod]
#     plt.scatter(df_subset["pl_bmassj"], df_subset["pl_eqt"])

# plt.legend()
# plt.show()

methods = small_df['discoverymethod'].unique()

for method in methods:
    subset = small_df[small_df['discoverymethod'] == method]
    plt.scatter(subset['pl_bmassj'], subset['pl_eqt'],
                label=method)

plt.xlabel('Planet Mass (Jupiter Masses)')
plt.ylabel('Equilibrium Temperature (K)')
plt.title('Planet Mass vs. Equilibrium Temperature by Discovery Method')

plt.legend()
plt.show()
```

```python
methods = small_df['discoverymethod'].unique()

for method in methods:
    subset = small_df[small_df['discoverymethod'] == method]
    plt.scatter(np.log(subset['pl_bmassj']), np.log(subset['pl_eqt']),
                label=method)

plt.xlabel('Planet Mass (Jupiter Masses)')
plt.ylabel('Equilibrium Temperature (K)')
plt.title('Planet Mass vs. Equilibrium Temperature by Discovery Method')

plt.legend()
plt.show()
```

```python
methods = small_df['discoverymethod'].unique()

plt.figure(figsize=(8, 6))

for method in methods:
    subset = small_df[small_df['discoverymethod'] == method]
    plt.scatter(np.log(subset['pl_bmassj']), np.log(subset['pl_eqt']),
                label=method)

coldest_planet = small_df[small_df['pl_bmassj'] > 5].sort_values('pl_eqt').iloc[0]

plt.scatter(np.log(coldest_planet['pl_bmassj']), np.log(coldest_planet['pl_eqt']),
            color='red', marker='*', s=200, edgecolor='black',
            label=f"Coldest: {coldest_planet['pl_name']}")

plt.xlabel('log(Planet Mass) (Jupiter Masses)')
plt.ylabel('log(Equilibrium Temperature) (K)')
plt.title('Planet Mass vs. Equilibrium Temperature (Log Scale)')

plt.legend()
plt.show()
```
