```python
# 3.2 2D NumPy Arrays
```

```python
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])
print(matrix.shape)
print(matrix.shape[0])
print(matrix.shape[1])
```

```python
print(matrix[:, 2])
```

```python
print(matrix[1])
```

```python
matrix_times2 = matrix * 2
print(matrix_times2)
```

```python
mean_colms = np.mean(matrix, axis=0)
print(mean_colms)
```

```python
overall_mean = np.mean(matrix)
print(overall_mean)
```

```python
max_each_row = np.max(matrix, axis=1)
print(max_each_row)
```

```python
min_each_colm = np.min(matrix, axis=0)
print(min_each_colm)
```

```python
flat = matrix.flatten()
print(flat)
print(flat.shape)
```

```python
# the 2D array was converted to a 1D array/list
```

```python
min_index = np.argmin(flat)
max_index = np.argmax(flat)
print(min_index)
print(max_index)
```

```python
min_coords = np.unravel_index(min_index, matrix.shape)
max_coords = np.unravel_index(max_index, matrix.shape)
print(min_coords)
print(max_coords)
print(matrix[min_coords])
print(matrix[max_coords])
```

```python
greater_than_15 = matrix[matrix > 15]
print(greater_than_15)
```

```python
sorted_matrix = np.sort(matrix, axis=1)
print(sorted_matrix)
```

```python
# 4 Prime Star Systems 
```

```python
arr = np.array([ 
    [2, 3, 5], 
    [14, 6, 8], 
    [11, 13, 17], 
    [7, 10, 13] 
])
```

```python
n = arr.max()
sieve = np.ones(n + 1, dtype=bool)
sieve[:2] = False  # 0 and 1 are not prime
for i in range(2, int(np.sqrt(n)) + 1):
    if sieve[i]:
        sieve[i*i:n+1:i] = False

primes = np.nonzero(sieve)[0]
prime_mask = np.isin(arr, primes)

rows_with_prime = np.any(prime_mask, axis=1)
result = arr[rows_with_prime]

print(result)
```

```python
# 5 Let's Play Checkers!
```

```python
import matplotlib.pyplot as plt
```

```python
checkerboard = np.zeros((8, 8), dtype=int)
checkerboard[1::2, ::2] = 1
checkerboard[::2, 1::2] = 1

reverse_checkerboard = np.zeros((8, 8), dtype=int)
reverse_checkerboard[::2, ::2] = 1
reverse_checkerboard[1::2, 1::2] = 1

fig, axes = plt.subplots(1, 2, figsize=(8, 4))

axes[0].imshow(checkerboard, cmap='gray', interpolation='none')
axes[0].set_title("Checkerboard")
axes[0].axis('off')

axes[1].imshow(reverse_checkerboard, cmap='gray', interpolation='none')
axes[1].set_title("Reverse Checkerboard")
axes[1].axis('off')

plt.show()
```
